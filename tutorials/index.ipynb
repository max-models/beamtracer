# Examples

The examples of every section, executed when the documentation is built. Download a notebook from its page to run it yourself.

**[Beam optics](/beamtracer/beam-optics/)**

- [Graded mesh](/beamtracer/examples/graded-mesh/) extracts a beam through an aperture and lets it drift on a mesh graded along the beam axis (`GridParams(x_spacing=...)`): the divergence of a uniform mesh with its fine cells everywhere, with half its nodes.
- [Mesh refinement](/beamtracer/examples/mesh-refinement/) extracts the beam on a coarse mesh with a refined box (`RefinementParams`) over the meniscus and the aperture.
- [Graded beam axis](/beamtracer/example-scripts/graded-beam-axis/) compares graded, refined and uniform meshes on a planar extraction and a long drift.
- [Electrostatic quadrupole](/beamtracer/examples/quadrupole-3d/) traces a proton beam through a quadrupole on a 3D grid, the case of IBSimu's `sim_3d_quad.py`.
- [Relativistic electrons](/beamtracer/example-scripts/relativistic-electrons/) accelerates electrons to 1 MeV and lets them gyrate in a magnetic field.

**[Charge changes](/beamtracer/charge-changes/)**

- [Stripping in the accelerator](/beamtracer/examples/accelerator-stripping/) adds the gas between the grids to an H⁻ beamlet: the share stripped along the accelerator, and the energy and divergence of the atoms it leaves (`mode="analog"`).
- [Neutralizer](/beamtracer/examples/neutralizer/) traces a 1 MeV D⁻ beam through a gas neutralizer channel coupled to the plasma it makes: the charge states at the exit and what the walls take from a diverging beam.

**[Beamline](/beamtracer/beamline/)**

- [Grid](/beamtracer/examples/beamline-grid/) turns a beamlet traced in 2D into 3D markers and copies it over an aperture grid with its aiming.
- [Dump](/beamtracer/examples/beamline-dump/) pushes the ions left after the neutralizer onto the plates of the residual-ion dump, and the reionized ions onto the duct walls in the stray field.
- [Duct](/beamtracer/examples/beamline-duct/) traces the neutrals through the duct to the port: the walls hit, the footprint, the reionization in the duct gas.
- [Calorimeter](/beamtracer/examples/beamline-calorimeter/) records the power map of a beamlet group and infers the beam's width and divergence from it as the experiments do.
- [Whole beamline](/beamtracer/examples/beamline/) puts the pieces in a row, from the grid to the port, and hands the neutrals on to freeofcharge.

**[Design optimisation](/beamtracer/optimisation/)**

- [Einzel lens](/beamtracer/examples/einzel-lens-optimization/) finds the voltage, gap and aperture of an einzel lens, against a scan and the thin-lens estimate.
- [Einzel lens design map](/beamtracer/example-scripts/einzel-lens-design-map/) maps the same lens over two parameters and draws the optimiser's path over the map.
- [Negative-ion beamlet](/beamtracer/examples/negative-ion-beamlet-optimization/) searches the extraction voltage, the aperture and the gap of an H⁻ beamlet for the smallest divergence.
