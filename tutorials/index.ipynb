# Examples

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Optimising an einzel lens](/beamtracer/examples/einzel-lens-optimization/) finds the voltage, gap and aperture of an einzel lens with `beamtracer.optimization`, against a scan and the thin-lens estimate.
- [Optimising a negative-ion beamlet](/beamtracer/examples/negative-ion-beamlet-optimization/) searches the extraction voltage, the EG aperture and the PG-EG gap of an H⁻ beamlet for the smallest divergence, on a coarse mesh checked against the example's.
- [The duct](/beamtracer/examples/beamline-duct/) traces the neutrals from the end of the traced region through the duct to the port with `beamtracer.beamline`: the walls hit, the footprint at the port, the reionization in the duct gas, against the closed forms.
- [The grid](/beamtracer/examples/beamline-grid/) turns a beamlet traced in 2D into 3D markers with `beamline.beamlet_markers` and copies it over an aperture grid with its aiming (`ApertureGrid`): the footprints of a flat and a curved grid along the beamline.
- [The dump](/beamtracer/examples/beamline-dump/) pushes the ions left after the neutralizer through the residual-ion dump onto its plates with `beamline.FieldRegion`, and bends the ions reionized in the duct onto its walls in the stray field.
- [The calorimeter](/beamtracer/examples/beamline-calorimeter/) records the power map of a beamlet group on a calorimeter with `beamline.Calorimeter` and infers the beam's width and divergence from it as the experiments do, against the markers' own divergence.
- [The beamline](/beamtracer/examples/beamline/) puts the pieces of `beamtracer.beamline` in a row, from an aimed group of beamlets through the dump and the duct to the port: where the power goes, the footprint at the port, and the neutral markers handed on to freeofcharge.
- [Scripts](/beamtracer/example-scripts/) are complete Python scripts with their plots.
- [Building the docs](/beamtracer/development/docs/) explains how to add an example and control its output.
